## Tensor Operations

extract max:

```
t = torch.rand(2, 2)
t
```

```
t.max()
```

```
t.max().item()
```

change type:

```
long_tensor = torch.tensor([[0,0,1],[1,1,1],[0,0,0]])
long_tensor.type()
```

```
float_tensor = torch.tensor([[0,0,1],[1,1,1],[0,0,0]]).to(dtype=torch.float32)
float_tensor.type()
```

functions starting with _ change tensor in-place:

```
random_tensor = torch.rand(2, 2)
print(random_tensor)
print(random_tensor.log2())
print(random_tensor)
```

```
print(random_tensor.log2_())
print(random_tensor)
```

reshaping for example MNIST (handwritings images) from 784 to 1(grascale) * 28 * 28(pixels):

```
flat_tensor = torch.rand(784)
view_tensor = flat_tensor.view(1,28,28) # view operates as a view of tensor.
# If the data changes, view also (unless new block of memory is allocated - call .contiguous() to fix)
view_tensor.shape
```

```
reshaped_tensor = flat_tensor.reshape(1,28,28) # preffered, as it performs contiguous behind the scenes
reshaped_tensor.shape
```

rearrange dimensions:

```
height_width_channel = torch.rand(640, 480, 3)
channel_height_width = height_width_channel.permute(2, 0, 1)
channel_height_width.shape
```

## Tensor Broadcasting

Broadcasting allows perform operations between a tensor and a smaller tensor. If
- Two dimencions are equal
- One dimencions is 1

```
torch.ones(2,2) + torch.ones(3,3)
```

```
torch.ones(1,3) + torch.ones(3,3)
```

```
torch.ones(3,3) + torch.ones(1,3)
```
